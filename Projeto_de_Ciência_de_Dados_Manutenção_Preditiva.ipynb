```python
# Recarregar a base de dados

!pip -q install ucimlrepo

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from ucimlrepo import fetch_ucirepo

dataset = fetch_ucirepo(id=601)

df = pd.concat(
    [dataset.data.features, dataset.data.targets],
    axis=1
)

print("Base carregada novamente!")
print("Dimensões:", df.shape)

display(df.head())
```

```python
variaveis = [
    "Air temperature",
    "Process temperature",
    "Rotational speed",
    "Torque",
    "Tool wear"
]

comparacao = df.groupby("Machine failure")[variaveis].mean().round(2)

comparacao.index = ["Sem falha", "Com falha"]

print("MÉDIAS DAS VARIÁVEIS POR GRUPO")
display(comparacao)

plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x="Machine failure", y="Torque")
plt.title("Torque por ocorrência de falha")
plt.xlabel("Falha (0 = não, 1 = sim)")
plt.ylabel("Torque")
plt.show()
```

```python
from sklearn.model_selection import train_test_split

# Características utilizadas para prever falhas
colunas_entrada = [
    "Type",
    "Air temperature",
    "Process temperature",
    "Rotational speed",
    "Torque",
    "Tool wear"
]

# X contém as características das máquinas
X = df[colunas_entrada].copy()

# y contém a resposta que queremos prever
y = df["Machine failure"].copy()

# Separar os dados em treino e teste
X_treino, X_teste, y_treino, y_teste = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Preparação concluída!")
print("Registros para treinamento:", len(X_treino))
print("Registros para teste:", len(X_teste))

print("\nFalhas no conjunto de treinamento:")
print(y_treino.value_counts())

print("\nFalhas no conjunto de teste:")
print(y_teste.value_counts())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    average_precision_score
)

# Separar características numéricas e categóricas
colunas_numericas = [
    "Air temperature",
    "Process temperature",
    "Rotational speed",
    "Torque",
    "Tool wear"
]

colunas_categoricas = ["Type"]

# Preparação das variáveis numéricas
preparo_numerico = Pipeline([
    ("imputacao", SimpleImputer(strategy="median")),
    ("escala", StandardScaler())
])

# Preparação das variáveis categóricas
preparo_categorico = Pipeline([
    ("imputacao", SimpleImputer(strategy="most_frequent")),
    ("codificacao", OneHotEncoder(handle_unknown="ignore"))
])

# Aplicar os tratamentos adequados a cada grupo de colunas
preprocessador = ColumnTransformer([
    ("numericas", preparo_numerico, colunas_numericas),
    ("categoricas", preparo_categorico, colunas_categoricas)
])

# Criar os quatro modelos
modelos = {
    "Baseline": DummyClassifier(strategy="most_frequent"),

    "Regressão Logística": LogisticRegression(
        class_weight="balanced",
        max_iter=2000,
        random_state=42
    ),

    "Árvore de Decisão": DecisionTreeClassifier(
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )
}

# Treinar cada modelo e avaliar no conjunto de teste
resultados = []
modelos_treinados = {}

for nome, algoritmo in modelos.items():

    pipeline = Pipeline([
        ("preparacao", preprocessador),
        ("modelo", algoritmo)
    ])

    pipeline.fit(X_treino, y_treino)

    previsoes = pipeline.predict(X_teste)

    # Probabilidade estimada para a classe de falha
    probabilidades = pipeline.predict_proba(X_teste)[:, 1]

    resultados.append({
        "Modelo": nome,
        "Acurácia": accuracy_score(y_teste, previsoes),
        "Precisão": precision_score(
            y_teste, previsoes, zero_division=0
        ),
        "Recall": recall_score(
            y_teste, previsoes, zero_division=0
        ),
        "F1-score": f1_score(
            y_teste, previsoes, zero_division=0
        ),
        "PR-AUC (AP)": average_precision_score(
            y_teste, probabilidades
        )
    })

    modelos_treinados[nome] = pipeline

# Apresentar os resultados
tabela_resultados = pd.DataFrame(resultados).set_index("Modelo")

display(tabela_resultados.round(3))
```

```python
from sklearn.model_selection import (
    RandomizedSearchCV,
    StratifiedKFold
)
from sklearn.base import clone

# Dividimos os dados de treinamento em 5 partes para validação cruzada.
validacao = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Configurações que serão testadas para cada algoritmo.
parametros = {
    "Regressão Logística": {
        "modelo__C": [0.01, 0.1, 1, 10, 100]
    },
    "Árvore de Decisão": {
        "modelo__max_depth": [None, 5, 10, 15, 20],
        "modelo__min_samples_split": [2, 5, 10, 20],
        "modelo__min_samples_leaf": [1, 2, 4, 8]
    },
    "Random Forest": {
        "modelo__n_estimators": [100, 200, 300],
        "modelo__max_depth": [None, 5, 10, 20],
        "modelo__min_samples_split": [2, 5, 10],
        "modelo__min_samples_leaf": [1, 2, 4],
        "modelo__max_features": ["sqrt", "log2"]
    }
}

# Usamos os modelos que já foram definidos na etapa anterior.
buscas = {}
resumo_buscas = []

for nome in parametros:
    pipeline = Pipeline([
        ("preparacao", clone(preprocessador)),
        ("modelo", clone(modelos[nome]))
    ])

    busca = RandomizedSearchCV(
        estimator=pipeline,
        param_distributions=parametros[nome],
        n_iter=10,
        scoring="average_precision",
        cv=validacao,
        random_state=42,
        n_jobs=-1,
        refit=True
    )

    busca.fit(X_treino, y_treino)
    buscas[nome] = busca

    resumo_buscas.append({
        "Modelo": nome,
        "PR-AUC médio na validação": busca.best_score_,
        "Melhores parâmetros": str(busca.best_params_)
    })

    print(f"\nModelo: {nome}")
    print(f"Melhor PR-AUC na validação: {busca.best_score_:.3f}")
    print("Melhores parâmetros:", busca.best_params_)

display(pd.DataFrame(resumo_buscas).round(3))
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    average_precision_score
)

resultados_ajustados = []

for nome, busca in buscas.items():
    melhor_modelo = busca.best_estimator_

    previsoes = melhor_modelo.predict(X_teste)
    probabilidades = melhor_modelo.predict_proba(X_teste)[:, 1]

    resultados_ajustados.append({
        "Modelo": nome,
        "Acurácia": accuracy_score(y_teste, previsoes),
        "Precisão": precision_score(
            y_teste, previsoes, zero_division=0
        ),
        "Recall": recall_score(
            y_teste, previsoes, zero_division=0
        ),
        "F1-score": f1_score(
            y_teste, previsoes, zero_division=0
        ),
        "PR-AUC (AP)": average_precision_score(
            y_teste, probabilidades
        )
    })

tabela_ajustada = (
    pd.DataFrame(resultados_ajustados)
    .set_index("Modelo")
)

display(tabela_ajustada.round(3))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

for nome, busca in buscas.items():
    modelo_final = busca.best_estimator_
    previsoes = modelo_final.predict(X_teste)

    cm = confusion_matrix(y_teste, previsoes)

    print(f"\nModelo: {nome}")
    print("Matriz de confusão:")
    print(cm)
    print("Ordem: [[verdadeiros negativos, falsos positivos],")
    print("        [falsos negativos, verdadeiros positivos]]")

    ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=["Sem falha", "Com falha"]
    ).plot(values_format="d")

    plt.title(f"Matriz de confusão — {nome}")
    plt.show()
```
